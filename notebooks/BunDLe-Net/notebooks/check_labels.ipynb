```python
import numpy as np
from data_reader import DataReader

# Abtastrate des Raw-Objekts, aus dem events_arr stammt
raw, _, _ = DataReader().read_data(subject="101")
sampling_rate_raw = raw.info["sfreq"]
del raw

events_arr = preprocessor.events_arr

is_image_stimulus = (events_arr[:, 2] >= 1) & (events_arr[:, 2] <= 64)
image_onsets_in_samples = events_arr[is_image_stimulus, 0]

gaps_in_seconds = np.diff(image_onsets_in_samples) / sampling_rate_raw

print(f"Abtastrate Raw: {sampling_rate_raw} Hz")
print(f"Anzahl Bilder: {len(image_onsets_in_samples)}")
print(f"Minimum: {gaps_in_seconds.min():.3f} s")
print(f"Median:  {np.median(gaps_in_seconds):.3f} s")
print(f"Maximum: {gaps_in_seconds.max():.3f} s")
print(f"Abstände unter 0.8 s: {np.sum(gaps_in_seconds < 0.8)} von {len(gaps_in_seconds)}")
```
