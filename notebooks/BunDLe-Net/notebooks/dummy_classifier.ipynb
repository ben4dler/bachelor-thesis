```python
import sys
sys.path.append("..")
from sklearn.dummy import DummyClassifier
from processed_data_storage import ProcessedDataStroage
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

subjects_range = range(1, 8)

processed_data_storage = ProcessedDataStroage()

processed_data_storage.preprocess_and_save_subjects_from_range(subjects_range)
X_train, X_test, y_train, y_test, category_per_epoch_all_subs_train, category_per_epoch_all_subs_test = processed_data_storage.read_subjects_from_range(subjects_range)

print(X_train.shape)
print(category_per_epoch_all_subs_train.shape)

preprocessor = processed_data_storage.preprocessor
train_all_windows_all_epochs = preprocessor.build_window_features(X_train, preprocessor.WINDOW_LENGTH)
test_all_windows_all_epochs = preprocessor.build_window_features(X_test, preprocessor.WINDOW_LENGTH)
print(f"Rohfenster Training: {train_all_windows_all_epochs.shape}")
print(f"Rohfenster Test: {test_all_windows_all_epochs.shape}")

strategies = ['most_frequent', 'stratified', 'uniform']

test_scores = []
for s in strategies:
    if s =='constant':
        dclf = DummyClassifier(strategy = s, random_state = 0, constant ='M')
    else:
        dclf = DummyClassifier(strategy = s, random_state = 0)
    dclf.fit(train_all_windows_all_epochs, category_per_epoch_all_subs_train)
    score = dclf.score(X_test, category_per_epoch_all_subs_test)
    test_scores.append(score)

ax = sns.stripplot(x=strategies, y=test_scores)  # Use x and y as keyword arguments
ax.set(xlabel='Strategy', ylabel='Test Score')
plt.show()
```
